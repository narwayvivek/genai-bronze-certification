**Write a Python program to draw (visualize) the architecture of a Neural Network used to classify fraudulent and non-fraudulent credit card transactions.**

**Assume the fraud detection dataset contains the following input features:**

**TransactionAmount**  
**TransactionTime**  
**MerchantCategory**  
**CustomerAge**  
**AccountBalance**  
**NumberOfTransactionsToday**  
**Fraud (0 = Genuine, 1 = Fraud)**


```python
# Credit Card Fraud Detection Neural Network
# Features -> Fraud classification (0=Genuine, 1=Fraud)

import matplotlib.pyplot as plt
import networkx as nx
import numpy as np

print("CREDIT CARD FRAUD DETECTION NEURAL NETWORK")
print("Features: TransactionAmount, Time, Merchant, Age, Balance, Transactions -> Fraud")

# Neural Network Architecture
G = nx.DiGraph()

# 1. INPUT LAYER (6 features exactly as specified)
features = ['Amount', 'Time', 'Mcat', 'Age', 'Bal', 'Trans']
input_nodes = [f'I{i}' for i in range(6)]
for i, feat in enumerate(features):
    G.add_node(input_nodes[i], label=feat)

# 2. HIDDEN LAYERS (Industry standard for fraud detection)
h1_nodes = [f'H1_{i}' for i in range(24)]  # 24 neurons
h2_nodes = [f'H2_{i}' for i in range(16)]  # 16 neurons
h3_nodes = [f'H3_{i}' for i in range(8)]   # 8 neurons

# 3. OUTPUT LAYER
G.add_node('Fraud', label='Fraud\n0/1')

# Add all nodes
for layer in [h1_nodes, h2_nodes, h3_nodes]:
    for node in layer:
        G.add_node(node)

# 4. FULLY CONNECTED (Standard NN)
layers = [input_nodes, h1_nodes, h2_nodes, h3_nodes, ['Fraud']]
for i in range(len(layers)-1):
    for n1 in layers[i]:
        for n2 in layers[i+1]:
            G.add_edge(n1, n2)

# 5. PROFESSIONAL POSITIONS
pos = {}
layer_x = [-4.5, -2.2, 0.3, 2.8, 5.2]
layer_sizes = [6, 24, 16, 8, 1]

for i, x_pos in enumerate(layer_x):
    size = layer_sizes[i]
    for j, node in enumerate(layers[i]):
        if size == 1:
            y_pos = 0
        else:
            y_pos = 1.4 - (j/(size-1)) * 2.6
        pos[node] = (x_pos, y_pos)

# 6. HIGH-QUALITY VISUALIZATION
plt.figure(figsize=(16, 12))

# Input layer (Blue)
nx.draw_networkx_nodes(G, pos, nodelist=input_nodes, node_color='#2E86C1',
                      node_size=1000, edgecolors='navy', linewidths=2.5)

# Hidden layers (Gradient orange->purple)
nx.draw_networkx_nodes(G, pos, nodelist=h1_nodes, node_color='#F18F01', node_size=650)
nx.draw_networkx_nodes(G, pos, nodelist=h2_nodes, node_color='#C73E1D', node_size=650)
nx.draw_networkx_nodes(G, pos, nodelist=h3_nodes, node_color='#8E44AD', node_size=650)

# Output layer (Red)
nx.draw_networkx_nodes(G, pos, nodelist=['Fraud'], node_color='#E74C3C',
                      node_size=1600, edgecolors='darkred', linewidths=4)

# Professional edges
nx.draw_networkx_edges(G, pos, alpha=0.35, arrowsize=12,
                      arrowstyle='->', width=0.9, edge_color='gray')

# Clean labels
labels = {n: G.nodes[n].get('label', n[-2:]) for n in G.nodes()}
nx.draw_networkx_labels(G, pos, labels, font_size=9, font_weight='bold')

# Layer headers with feature names
plt.text(-4.5, 2.4, 'INPUT LAYER\n6 Features', fontsize=13, fontweight='bold',
         ha='center', bbox=dict(boxstyle='round,pad=0.4', facecolor='#2E86C1', alpha=0.8))
plt.text(-2.2, 2.4, 'HIDDEN 1\n24 Neurons\n(ReLU)', fontsize=13, fontweight='bold',
         ha='center', bbox=dict(boxstyle='round,pad=0.4', facecolor='#F18F01', alpha=0.8))
plt.text(0.3, 2.4, 'HIDDEN 2\n16 Neurons\n(ReLU)', fontsize=13, fontweight='bold',
         ha='center', bbox=dict(boxstyle='round,pad=0.4', facecolor='#C73E1D', alpha=0.8))
plt.text(2.8, 2.4, 'HIDDEN 3\n8 Neurons\n(ReLU)', fontsize=13, fontweight='bold',
         ha='center', bbox=dict(boxstyle='round,pad=0.4', facecolor='#8E44AD', alpha=0.8))
plt.text(5.2, 2.4, 'OUTPUT\nFraud\n(Sigmoid)', fontsize=13, fontweight='bold',
         ha='center', bbox=dict(boxstyle='round,pad=0.4', facecolor='#E74C3C', alpha=0.8))

# Title and polish
plt.title('CREDIT CARD FRAUD DETECTION NEURAL NETWORK ARCHITECTURE\n'
          'TransactionAmount → TransactionTime → MerchantCategory → CustomerAge → AccountBalance → NumberOfTransactionsToday → Fraud(0/1)',
          fontsize=16, fontweight='bold', pad=30)

plt.axis('off')
plt.tight_layout()
plt.show()

print("\n✅ Neural Network Architecture Complete!")
print("📋 Features used: TransactionAmount, TransactionTime, MerchantCategory, CustomerAge, AccountBalance, NumberOfTransactionsToday")
print("🎯 Architecture: 6 → 24 → 16 → 8 → 1 (Fully Connected)")

```
